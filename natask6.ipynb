```python
products = products.drop_duplicates()
customers = customers.drop_duplicates()
```

```python
sales = orders.merge(products, on="ProductID", how="left")
sales = sales.merge(customers, on="CustomerID", how="left")
```

```python
sales.shape
```

```python
sales.head()
```

```python
sales["OrderDate"] = pd.to_datetime(sales["OrderDate"])
sales
```

```python
sales["Month"] = sales["OrderDate"].dt.month
sales
```

```python
monthly_sales = sales.groupby("Month")["SalesAmount"].sum()
sales
```

```python
plt.figure(figsize=(10,5))
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid()
plt.show()
```

```python
top_products = sales.groupby("ProductName")["SalesAmount"].sum().nlargest(10)
sales
```

```python
plt.figure(figsize=(10,5))
top_products.plot(kind="bar", color="teal")

plt.title("Top 10 Products by Sales")
plt.xlabel("Product Name")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_sales = sales.groupby("Category")["SalesAmount"].sum().sort_values(ascending=False)
sales
```

```python
plt.figure(figsize=(8,5))
category_sales.plot(kind="bar", color="coral")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
segment_sales = sales.groupby("Segment")["SalesAmount"].sum().sort_values(ascending=False)
sales
```

```python
plt.figure(figsize=(8,5))
segment_sales.plot(kind="bar", color="skyblue")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
state_sales = sales.groupby("State")["SalesAmount"].sum().nlargest(10)
sales
```

```python
plt.figure(figsize=(10,5))
state_sales.plot(kind="bar", color="orange")

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
channel_sales = sales.groupby("SalesChannel")["SalesAmount"].sum().sort_values(ascending=False)
sales
```

```python
plt.figure(figsize=(8,5))
channel_sales.plot(kind="bar", color="mediumseagreen")

plt.title("Sales by Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_count = sales["PaymentMethod"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(payment_count, labels=payment_count.index,
        autopct="%1.1f%%", startangle=90)

plt.title("Payment Method Distribution")
plt.show()
```

```python
status_count = sales["OrderStatus"].value_counts()

plt.figure(figsize=(8,5))
status_count.plot(kind="bar", color="mediumpurple")

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales["ProfitMargin"] = (
    sales["Profit"] / sales["SalesAmount"].replace(0, np.nan)
) * 100

print("Average Profit Margin:", sales["ProfitMargin"].mean())

discount_profit = sales.groupby("DiscountPct")["Profit"].mean()
```

```python
plt.figure(figsize=(9,5))
discount_profit.plot(kind="line", marker="o")

plt.title("Discount Percentage vs Average Profit")
plt.xlabel("Discount Percentage")
plt.ylabel("Average Profit")
plt.grid()
plt.show()
```

```python
sns.scatterplot(
    data=sales,
    x='DiscountPct',
    y='Profit',
    hue='Category'
)

plt.title('Discount vs Profit')
plt.xlabel('Discount Percentage')
plt.ylabel('Profit')
plt.show()
```
